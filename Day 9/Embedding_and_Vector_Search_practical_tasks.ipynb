```python
from sentence_transformers import SentenceTransformer
```

Load the 'paraphrase-MiniLM-L6-v2' model  from HuggingFace resource  using the  SentenceTransformer( *model-name* )  and store the reference to the model object in the 'model' variable

```python
# place your code here
model = SentenceTransformer('paraphrase-MiniLM-L6-v2')
```

After loading the model, call the 'encode()' method on the model object to create a vector representation of a specific text sentence. Use your own text string  as the parameter.

```python
# complete the code
sentence = "Kaunas University of Technology provides several courses for Bachelors and Masters in the field of Computer Science"  #your sentence
embedding = model.encode(sentence)
print(embedding)
print("Shape:", embedding.shape)
```

Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python
# complete the code
sentences_list = ["Kaunas University of Technology provides several courses for bachelor's and master's students in the field of computer science.",
    "Students at KTU can choose different study programmes related to software engineering, artificial intelligence, and information technology.",
    "The university offers opportunities for students to participate in research projects and gain practical experience during their studies.",
    "Computer science students learn programming languages, algorithms, databases, and software development as part of their academic education.",
    "Many international students choose Kaunas University of Technology because of its English taught programmes and multicultural learning environment.",
    "KTU students can take part in internships and collaborate with companies to gain experience before graduating from university.",
    "The university provides bachelor's and master's degree programmes in engineering, business, technology, and other academic disciplines.",
    "Students studying computer science often work on practical projects involving artificial intelligence, machine learning, and software development."
]    #your sentences

sentence_embeddings = model.encode(sentences_list)

print("Number of sentences:", len(sentences_list))
print("Shape:", sentence_embeddings.shape)

```

#**Definition of semantic textual similaritye**

Import 'util' module from sentence_transformers library.

You can calculate the cosine similarity of the vector representations of our sentences using the 'cos_sim()' function from the util module.
Example: sim = util.cos_sim(embedding_1, embedding_2). Calculate the cosine similarity for any two sentences from your list.


```python
# place your code here
similarity = util.cos_sim(
    sentence_embeddings[0],
    sentence_embeddings[1]
)

print("Similarity:", similarity.item())
```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
# place your code here
def cos_similarity_calculation(text):
    text_embedding = model.encode(text)

    scores = util.cos_sim(text_embedding, sentence_embeddings)

    for i, score in enumerate(scores[0]):
        print(f"Sentence {i + 1}: {score.item():.4f}")
        print(sentences_list[i])
```

```python
# testing the function
cos_similarity_calculation(
    "I want to study artificial intelligence in Kaunas and gain practical experience through university projects"
)
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
# place your code here
import numpy as np

def cosine_similarity_batch(vector, batch_vectors):
    dot_products = np.dot(batch_vectors, vector)

    vector_norm = np.linalg.norm(vector)
    batch_norms = np.linalg.norm(batch_vectors, axis=1)

    similarities = dot_products / (vector_norm * batch_norms)

    return similarities
```

```python
query_sentence = "I want to study artificial intelligence in Kaunas and gain practical experience through university projects."

query_vector = model.encode(query_sentence)

scores = cosine_similarity_batch(query_vector, sentence_embeddings)

for i, score in enumerate(scores):
    print(f"Sentence {i + 1}: {score:.4f}")
    print(sentences_list[i])
    print()
```
