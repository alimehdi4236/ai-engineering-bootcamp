#### 3. Call the OpenAI API with different `temperature` values

Prompt: `"Write a two sentence story about a robot learning to cook."`

`temperature` values: `0`, `0.7` and `1.5`. Generate three responses at each temperature.

Use `gpt-5.4-mini` for this task and set `reasoning_effort="none"`.

Compare the outputs.

Tip: With Chat Completions API you can use parameter `n=3` to generate three different responses to your prompt with one call.

````python
# with temp 0
prompt = "Write a two sentence story about a robot learning to cook."

response_t0 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}],
    temperature=0,
    reasoning_effort="none",
    n=3
)

print("Temperature: 0")
for i, choice in enumerate(response_t0.choices, 1):
    print(f"\nResponse {i}:")
    print(choice.message.content)

print("\nUsage:", response_t0.usage)
````

````python
#with temp 0.7

response_t07 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}],
    temperature=0.7,
    reasoning_effort="none",
    n=3
)

print("Temperature: 0.7")
for i, choice in enumerate(response_t07.choices, 1):
    print(f"\nResponse {i}:")
    print(choice.message.content)

print("\nUsage:", response_t07.usage)
````

````python
# with temp 1.5

response_t15 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}],
    temperature=1.5,
    reasoning_effort="none",
    n=3
)

print("Temperature: 1.5")
for i, choice in enumerate(response_t15.choices, 1):
    print(f"\nResponse {i}:")
    print(choice.message.content)

print("\nUsage:", response_t15.usage)
````

# comparison
outputs became more varied as i increased the temperature. At temp 0 second and third responses were exactly the same but first one was different. I think 0.7 gave a good balance between variation and keeping the story focused while 1.5 allowed more unusual ideas it showd me that temperature influences the randomness of token selection regardless of quality of response

#### 11. Optimize input tokens.

Prompt: `"As you know, streaming services and social media platforms use artificial intelligence to recommend movies, shows, and songs to people. These smart recommendation systems work by looking closely at your past viewing habits, what categories you like, and what similar users enjoy watching, and then they predict what you might want to see next. Keeping this background in mind, explain in two simple sentences why a user who loves sci-fi movies suddenly sees a romantic comedy recommended."`

Rewrite the prompt to use as few input tokens as possible while preserving its meaning and the requirement to answer in two simple sentences. The response should remain relevant, clear, and accurate.

Run the original and revised prompts using the same model and settings. Compare their input token counts and assess whether shortening the prompt affected the response quality.

You can use the API response’s usage information to check token counts.

````python
original_prompt = """As you know, streaming services and social media platforms use artificial intelligence to recommend movies, shows, and songs to people. These smart recommendation systems work by looking closely at your past viewing habits, what categories you like, and what similar users enjoy watching, and then they predict what you might want to see next. Keeping this background in mind, explain in two simple sentences why a user who loves sci-fi movies suddenly sees a romantic comedy recommended."""

response_original = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {"role": "user", "content": original_prompt}
    ],
    reasoning_effort="none"
)

print("Original prompt response:")
print(response_original.choices[0].message.content)

print("\nOriginal prompt tokens:", response_original.usage.prompt_tokens)
print("Completion tokens:", response_original.usage.completion_tokens)
````

````python
optimized_prompt = """In two simple sentences explain why a streaming service recommends a romantic comedy to a user who loves sci-fi considering their view history, preferences and what similar users enjoy."""

response_optimized = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {"role": "user", "content": optimized_prompt}
    ],
    reasoning_effort="none"
)

print("Optimized prompt response:")
print(response_optimized.choices[0].message.content)

print("\nOptimized prompt tokens:", response_optimized.usage.prompt_tokens)
print("Completion tokens:", response_optimized.usage.completion_tokens)
````

# Observatiion
The shorter prompt used 39 tokens instead of 94 which is a good reduction. I was expecting some information to be lost after removing so much text but both responses still follows the two sentence requirement and explains the recommendation clearly. The original answer focused on similar users and popular content while the optimized one also mentioned the users possible interest in character-driven stories. The wording changed a little but the main idea remained same. its a good example of how unnecessary background text can be removed without making the prompt less useful

````python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# Test cases
print(factorial(0))  # 1
print(factorial(1))  # 1
print(factorial(5))  # 120
print(factorial(7))  # 5040
````

Extract the information from the text above using two different methods:

Which method yields a more reliable result for a computer program to process, and why?

````python
input_text = """
Our trip to Tokyo starts on June 12th. We're staying for 5 nights at the Park Hyatt before heading to Kyoto.
"""

prompt = f"""
Extract the travel information from this text. Return a JSON object with exactly these fields:
destination: the first city mentioned
start_date: the starting date in YYYY-MM-DD format
stay_duration: number of nights as an integer
Important things to remember:
Do not include explanations or Markdown and dont invent missing information. return null if the year is unknown in start_date

Text: {input_text}
"""

response_prompt_json = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {"role": "user", "content": prompt}
    ],
    response_format={"type": "json_object"},
    reasoning_effort="none"
)

print(response_prompt_json.choices[0].message.content)
````

````python
import json

schema = {
    "type": "object",
    "properties": {
        "destination": {
            "type": "string"
        },
        "start_date": {
            "type": ["string", "null"],
            "format": "date"
        },
        "stay_duration": {
            "type": "integer"
        }
    },
    "required": [
        "destination",
        "start_date",
        "stay_duration"
    ],
    "additionalProperties": False
}

response_schema_json = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Extract the travel information from the user's text. "
                "The destination is the first city mentioned. "
                "Do not invent a year. Use null if the year is missing."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
    response_format={
        "type": "json_schema",
        "json_schema": {
            "name": "travel_information",
            "strict": True,
            "schema": schema
        }
    },
    reasoning_effort="none"
)

print(response_schema_json.choices[0].message.content)
````
